```python
inflation = pd.read_csv(
    "../Data/External/World Bank Inflation/API_FP.CPI.TOTL.ZG_DS2_en_csv_v2_446435.csv",
    skiprows=4
)
```

```python
inflation.head()
```

```python
inflation.columns.tolist()
```

```python
inflation.shape
```

```python
inflation = inflation.drop(
    columns=[
        "Country Code",
        "Indicator Name",
        "Indicator Code",
        "Unnamed: 70"
    ]
)
```

```python
inflation.head()
```

```python
inflation = inflation.melt(
    id_vars="Country Name",
    var_name="year",
    value_name="inflation_rate"
)
```

```python
inflation.head()
```

```python
inflation["year"] = inflation["year"].astype(int)
```

```python
min_year = retail["year"].min()
max_year = retail["year"].max()

inflation = inflation[
    inflation["year"].between(min_year, max_year)
]
```

```python
inflation.head()
```

```python
inflation.shape
```

```python
inflation = inflation.rename(
    columns={
        "Country Name": "country"
    }
)
```

```python
inflation.columns
```

```python
pd.merge(retail, inflation, on=["country", "year"])
```

```python
retail_inflation = pd.merge(
    retail,
    inflation,
    on=["country", "year"],
    how="left"
)

retail_inflation["inflation_rate"].isnull().sum()
```

```python
retail_inflation[
    retail_inflation["inflation_rate"].isnull()
]["country"].unique()
```

```python
inflation[
    inflation["country"].str.contains("Hong|Russia|Egypt|Iran|Korea|Viet|Vene|Congo|Czech|Slovak|Maced|Burma|Swaz|Taiwan", case=False, na=False)
]["country"].unique()
```

```python
country_mapping = {
    "Hong Kong": "Hong Kong SAR, China",
    "Democratic Republic of the Congo": "Congo, Dem. Rep.",
    "Republic of the Congo": "Congo, Rep.",
    "Egypt": "Egypt, Arab Rep.",
    "Iran": "Iran, Islamic Rep.",
    "South Korea": "Korea, Rep.",
    "North Korea": "Korea, Dem. People's Rep.",
    "Russia": "Russian Federation",
    "Vietnam": "Viet Nam",
    "Venezuela": "Venezuela, RB",
    "Macedonia": "North Macedonia",
    "Czech Republic": "Czechia"
}

retail["country"] = retail["country"].replace(country_mapping)
```

```python
retail_inflation = pd.merge(
    retail,
    inflation,
    on=["country", "year"],
    how="left"
)
```

```python
retail_inflation["inflation_rate"].isnull().sum()
```

```python
retail_inflation[
    retail_inflation["inflation_rate"].isnull()
]["country"].value_counts()
```
