```python
import pandas as pd
import numpy as np
```

```python
retail = pd.read_csv("../Data/Processed/cleaned_superstore.csv")

inflation = pd.read_csv("../Data/Processed/inflation_cleaned.csv")

gdp = pd.read_csv("../Data/Processed/gdp_growth_cleaned.csv")
```

```python
print("Retail:", retail.shape)
print("Inflation:", inflation.shape)
print("GDP:", gdp.shape)
```

```python
print(retail.columns.tolist())
print(inflation.columns.tolist())
print(gdp.columns.tolist())
```

```python
country_mapping = {
    "Hong Kong": "Hong Kong SAR, China",
    "Democratic Republic of the Congo": "Congo, Dem. Rep.",
    "Republic of the Congo": "Congo, Rep.",
    "Egypt": "Egypt, Arab Rep.",
    "Iran": "Iran, Islamic Rep.",
    "South Korea": "Korea, Rep.",
    "North Korea": "Korea, Dem. People's Rep.",
    "Russia": "Russian Federation",
    "Vietnam": "Viet Nam",
    "Venezuela": "Venezuela, RB",
    "Macedonia": "North Macedonia",
    "Czech Republic": "Czechia",
    "Turkey": "Turkiye",
    "Myanmar (Burma)": "Myanmar",
    "Kyrgyzstan": "Kyrgyz Republic",
    "Slovakia": "Slovak Republic",
    "Swaziland": "Eswatini"
}

retail["country"] = retail["country"].replace(country_mapping)

retail["country"] = retail["country"].replace({
    "Türkiye": "Turkiye"
})
```

```python
final_df = pd.merge(
    retail,
    inflation,
    on=["country", "year"],
    how="left"
)
```

```python
final_df = pd.merge(
    final_df,
    gdp,
    on=["country", "year"],
    how="left"
)
```
