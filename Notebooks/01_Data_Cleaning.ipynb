```python
import pandas as pd
```

```python
df = pd.read_csv("../Data/Raw/global_retail_raw.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df.columns = df.columns.str.strip()

df.columns = df.columns.str.lower()

df.columns = df.columns.str.replace(".", "_", regex=False)
```

```python
df.columns
```

```python
df["记录数"].value_counts()
```

```python
df.drop(columns=["记录数"], inplace=True)
```

```python
df["order_date"] = pd.to_datetime(df["order_date"])
```

```python
df["order_date"].dtype
```

```python
df["ship_date"] = pd.to_datetime(df["ship_date"])
```

```python
df["ship_date"].dtype
```

```python
df["delivery_days"] = (df["ship_date"] - df["order_date"]).dt.days
```

```python
import numpy as np

df["profit_margin"] = np.where(
    df["sales"] == 0,
    0,
    (df["profit"] / df["sales"]) * 100
)
```

```python
df["month"] = df["order_date"].dt.month_name()
```

```python
df["month_no"] = df["order_date"].dt.month
```

```python
df["quarter"] = df["order_date"].dt.quarter
```

```python
df["day"] = df["order_date"].dt.day
```

```python
df["year"] = df["order_date"].dt.year
```

```python
df["year"] = df["order_date"].dt.year
```

```python
df["week"] = df["order_date"].dt.isocalendar().week
```

```python
df["is_weekend"] = df["order_date"].dt.dayofweek >= 5
```

```python
df["discount_band"] = pd.cut(
    df["discount"],
    bins=[-0.01, 0, 0.1, 0.2, 0.3, 0.5, 1],
    labels=[
        "No Discount",
        "0-10%",
        "10-20%",
        "20-30%",
        "30-50%",
        "Above 50%"
    ]
)
```

```python
df["loss_order"] = (df["profit"] < 0).astype(int)
```

```python
df["high_discount"] = (df["discount"] >= 0.20).astype(int)
```

```python
df["sales_bucket"] = pd.qcut(
    df["sales"],
    q=4,
    labels=["Low", "Medium", "High", "Very High"]
)
```

```python
df["profit_bucket"] = pd.qcut(
    df["profit"],
    q=4,
    labels=["Low", "Medium", "High", "Very High"]
)
```

```python
df["shipping_cost_bucket"] = pd.qcut(
    df["shipping_cost"],
    q=4,
    labels=["Low", "Medium", "High", "Very High"]
)
```

```python
df["average_order_value"] = df["sales"] / df["quantity"]
```

```python
df["shipping_cost_pct"] = np.where(
    df["sales"] == 0,
    0,
    (df["shipping_cost"] / df["sales"]) * 100
)
```

```python
df["discount_amount"] = df["sales"] * df["discount"]
```

```python
df["net_sales"] = df["sales"] - df["discount_amount"]
```

```python
import numpy as np
```

```python
df["profit_status"] = np.where(
    df["profit"] > 0,
    "Profit",
    np.where(df["profit"] == 0, "Break Even", "Loss")
)
```

```python
df["delivery_status"] = np.where(
    df["delivery_days"] <= 3,
    "Fast",
    np.where(
        df["delivery_days"] <= 7,
        "Normal",
        "Delayed"
    )
)
```

```python
df["profit_per_unit"] = df["profit"] / df["quantity"]
```
